```python
from mqt.qmap.ph.data_collection import Setup, collect_pipeline_results, export_results_table
from mqt.qmap.ph.subcircuit_compilation import OptimizationConfig
```

```python
setups = [
    Setup(num_modes=24, target_dim=4),
    Setup(num_modes=48, target_dim=4),
    Setup(num_modes=24, target_dim=8),
    Setup(num_modes=48, target_dim=8),
]

phase_errors = [0.00, 0.015, 0.03]

config = OptimizationConfig(
    lr=0.01,
    threshold=1e-6,
    num_restarts=1,
    max_iterations=1000,
    restart_perturbation=0.15,
    exclude_edge_phase_shifters=False,
    optimize_routing_parameters=True,
)

df_results = collect_pipeline_results(
    setups=setups,
    config=config,
    num_unitaries_per_setup=4,
    repeats_per_unitary=1,
    phase_errors=phase_errors,
    base_seed=42,
    input_losses=True,
    output_losses=True,
    ideal_beam_splitters=False,
)

export_results_table(
    df_results,
    csv_path="results/pipeline_results.csv",
)

print(df_results.head())
print(f"Collected rows: {len(df_results)}")
```
